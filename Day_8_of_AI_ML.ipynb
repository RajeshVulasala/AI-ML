**1. DataFrames:**  
--> DataFrame is a `table data`.  
--> It contains an array of entries, each of which has certain values.  
--> Each entry corresponding to a a row and a column.

````python
pd.DataFrame({'yes':[50,21], 'No':[131, 2]})
````

DataFrames entries are not limited to integers

````python
pd.DataFrame({'bob':['I liked it.', 'It was awful.'],
              'sue':['Pretty good.', 'Bland.']})
````

To `Change index values` from 0,1,2,... in the DataFrame.

````python
pd.DataFrame({'bob':['I liked it.', 'It was awful.'],
              'sue':['Pretty good.', 'Bland.']},
              index = ['product A', 'product B'])
````

**2. SERIES:**


*   Sequence of data values
*   if DataFrame is Table, Then `SERIES is a LIST`



````python
# series
# code is same as DataFrames
pd.Series([1,2,3,4,5])
````



1. Series is a single column of a DataFrame. So, you can assign row lables to the series by using `index=` parameter.
2.  A series does not have a column name. It only has only one Overal name. we can define it using `name=` parameter.



````python
pd.Series([30,40,50], index=['2023 Sale', '2024 Sales', '2025 Sales'], name='Soaps')
````

## pd.read_csv()


*   It has over 30 optional perameters you can specify.
* Example:  you can see in this dataset that the CSV file has a index, which pandas pick up on automatically.  
To make pandas do not use that column for the index, we can specify an `index_col` parameter



````python
# Ex: index_col parameter
df = pd.read_csv('Attacks.pcap_ISCX.csv', index_col=0)
df.head()
````

As you can see the column of index is removed it means the csv file has no built in index column.

IF you want to convert you created data frame into CSV you can use this,  
`variablename.to_csv("Filename.csv")`

````python
# Ex: .to_csv()
df = pd.DataFrame({'cows':[12,20], 'Goats':[22,19]}, index=['Year 1', 'Year 2'])
df
````

````python
#  we can convert the above data frame into csv of filename: animals.csv
df.to_csv("animals.csv")
````
