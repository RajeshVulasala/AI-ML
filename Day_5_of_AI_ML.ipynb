**TUPLES**: same as lists

1. The syntx for creating them uses perantheses instead of square brackets.

````python
# Syntax EX:
x = (1,2,3)
y = 4, 5, 6
print(x)
print(y)
````

2. The TUPLES cannot be modified as list.
Tuples are **immutable**.

````python
# EX:
x[1]=2.5
````

Tuples are often used in functions with multiple return values.
1. Ex: .as_integer_ratio() --> method of float objects returns a numerator and a denominator in a form of tuple.

````python
# list.as_integer_ratio()
z = 0.125
z.as_integer_ratio()
````

````python
# Thse multiolr return values can be individually assigned as follows:/
numerator, denominator = z.as_integer_ratio()
print(numerator / denominator)
````
