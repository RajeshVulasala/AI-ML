# **Grouping**
* Maps allows us to transform the data in a dataframe and series. one value at a time for the entire column
* we of often want to grou the data and want to do something to the group data in it.
This is solved by the grouping function `groupby()`.

````python
df.groupby('points').points.count()
````

`groupby()` created a group of df which alloted the same point values

````python
df.groupby('points')
````

`df.groupby('points')` It groups the dataframe by the points columns. And then wait for us to give or perform some operations.  
For Example, Important operations:
1. `.count()`
2. `.mean()`
3. `.size()`
4. `.min()`

````python
df.groupby('points').price.min()
````

** You can think of each group we genarate as being a slice of our dataframe contaning only data with values that match.  
we can do that by using the `.apply()` method, and we can manipulate the data in any we see fit.

````python
# Ex:
df.groupby('winery').apply(lambda ndf: ndf.title.iloc[0])
````

`lambda ndf: ndf.title.iloc[0]` is nothong but an ananmous function:

## for more fine-grained control, we can also group by more tham one column.

Example: If we have to pick out the best wine by country and province.

````python
# EX:
df.groupby(['country', 'province']).apply(lambda ndf: ndf.loc[ndf.points.idxmax()])
````

# **.add()**
* It is another worthy `groupby()` method.
* which lets you run bunch of different functions on your DataFrame simulutaneiously.

EXAMPLE: we can genearate a simple summary of the dataset.

````python
# EX:
df.groupby('country').price.agg([len, min, max])
````

# Multi-Indexes  
* Multi index differes from normal index because it has multiple levels.  
EXAMPLE:  
|  |  | len   |
|-------|-----|----------|
| country | province  |  |
|  |  | |
|  |   | |

* It has the multi level indexs

````python
multi_indexes = df.groupby(['country', 'province']).description.agg([len])
````

````python
type(multi_indexes.index)
````

* Multi indeces have several methods for dealing with their tiered structure which are absent for sing-level indices.  
* They also require 2 levels of labels to retreive a value.
* In general multi-index method you eill use most often is the one for converting multi-index to single-index.


# `.reset_index()`

````python
multi_indexes.reset_index()
````

# **Sorting**
* To get the data in sorting order.
* we use `sort_values()` method to sort the data

````python
# EX:
Sorting = multi_indexes.reset_index()
Sorting.sort_values(by='len')
````

## By default lowest value go first in sorting value, If we want we can make higher value go first

````python
# EX:
Sorting.sort_values(by='len', ascending=False)
````

## We can sort by more than one calumn at a time in `sorting values`

````python
Sorting.sort_values(by=['country', 'len'])
````

# Sorting the data using Index by using `.sort_index()`

````python
Sorting.sort_index()
````
