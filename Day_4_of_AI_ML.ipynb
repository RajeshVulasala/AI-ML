Functions Applied to Functions
* can supply functions as arguments to other functions.

````python
def mult(y):
  return 5*y
def call(fn, arg):
  """Call fn on arg"""
  return fn(arg)
def squred_call(fn, arg):
  """Call fn on the result of calling fn on arg"""
  return fn(fn(arg))
print(
    call(mult,1),
    squred_call(mult,1),
    sep='\n'
)
````

MAX Function

````python
# EX:
def mod_5(X):
  return X % 5
print(
    'Which is the biggest number?',
    max(100, 51, 14),
    'Which is the biggest modulo 5?',
    max(100, 51, 14, key=mod_5),
    sep='\n',
)
````
