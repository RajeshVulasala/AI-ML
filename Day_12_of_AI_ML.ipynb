# Missing Data
* Entries missing values are given athe value `NaN` (Not a Number).
* For Technical reasons for all `NaN` values are always of the `float64` dtype.  
There are Specific methods to missing data. To select `NaN` entries we can use `pd.isnull()` or `pd.notnull()`

````python
# EX:
pd.isnull(df.country) # Here It checked wethere the country values NaN or not.
````

````python
df[pd.isnull(df.country)] # Here it changed the country values into NaN
````

# 2. Renaming and Combining

Data will come to us with column names and, index names, or other naming conventions that we are not satisfied with we can change them here.

## Renaming
* `rename()` It's let you change index names and/or column names.

````python
# EX:
df.rename(columns={'points': 'score'})
````

`rename()` also lets you rename the index or column values by specifying a `index` or `column` keyword parameter.

````python
# EX:
df.rename(index={0: 'firstEntry', 1: 'secondEntry'})
````

## Index name of Row and Column axies
here the both row index and column index can have their own attribute `name`.  
by using `rename_axis()`.


```
rename_axis()
```

````python
df.rename_axis("wines", axis='rows').rename_axis("fields", axis='columns')
````

# Combining
While performing operations on a dataset, we will sometines need to combine different DataFrames and Series in non-trival ways.

Pandas has `three` methods for doing this
1. concat()
2. join()
3. merge()  
Most of what `merge()` can do can also be done by `join()`.

# Here we are using two different DataFrames objects but having the same fields.  
**Data-Sets Used**
1. [CAvideos.csv](https://www.kaggle.com/datasets/datasnaek/youtube-new?select=CAvideos.csv)
2. [GBvideos.csv](https://www.kaggle.com/datasets/datasnaek/youtube-new?select=CAvideos.csv)

````python
candian_youtube = pd.read_csv("CAvideos.csv")
british_youtube = pd.read_csv("GBvideos.csv")
````

````python
pd.concat([candian_youtube, british_youtube])
````

#Concat()

````python
pd.concat([candian_youtube, british_youtube])
````

# join()
lets you combine different DataFrames objects which have an index in common

````python
left = candian_youtube.set_index(['title', 'trending_date'])
right = british_youtube.set_index(['title', 'trending_date'])

left.join(right, lsuffix='_CAN', rsuffix='_UK')
````
