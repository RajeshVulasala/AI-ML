**LOOPS AND LIST COMPREHESIONS**

**LOOPS**: It is way to execute a peace of code repeatedly

```python
# EX: 1
planets = ['Mercury', 'Venus', 'Earth', 'Mars', 'Jupiter', 'Saturn', 'Uranus', 'Neptune']
```

```python
# loop to print all planets in line
for i in planets:
  print(i, end=',')
```

```python
# EX: 2
multiples = (2, 4, 6, 8, 10, 12)
product = 1
for i in multiples:
  product = product * i
product
```

**range()**--> to repeat a action several times.

```python
for i in range(5):
  print("Doing important work. i =", i)
```

**WHILE LOOP**: Iterated until some condition is met

```python
i = 10;
while i > 0:
  print(i, end=' ')
  i -= 1
```

```python
i = 0;
while i < 10:
  print(i, end=' ')
  i += 1
```

The argument of WHILE loop is executed as soon as a boolean stmt, and the loop is executed until the stmt evaluate False

```python
i = 1
while i <=10:
  if i % 2 == 0:
    print(i,"= even")
  else:
    print(i, "= odd")
  i += 1
```

**LIST COMPREHENSIONS**: These are the most beloved unique features in python.

```python
# EX: 1
squares = [n**2 for n in range(10)]
squares
```

```python
# Same above example in different way
squares = []
for n in range(10):
    squares.append(n**2)
squares
```

We can also add an if stmt in list

```python
short_planets = [i for i in planets if len(i) < 6]
short_planets
```
