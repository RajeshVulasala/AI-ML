# MAPS
* For a function that takes one set of values and "maps" to another set of values.
* In data science we often have a need for creating new representations from exixting data, or for transforming data from the format it is now to the format that we want it to be in later.  


## There are 2 mapping methods that you will use often.

```python
df_mean = df.points.mean()
df.points.map(lambda p: p - df_mean)
```

## Explanation
df_mean = gives the the mean value of points column in df.  
.map() = applies function to each value in the column.  
p: = For every value of p.  
p - df_mean = for every p value do this p-df_mean

```python
# EX: for .map
sf = pd.DataFrame({'pts':[80,90,100]})
pts_mean = sf.pts.mean()
print(sf)
sf_mean = sf.pts.mean()
sf.pts.map(lambda p: p - sf_mean)
```

# How it works:
Now .map() runs:

| p (value) | Calculation	| Result  |
|-----------|-------------|---------|
| 80	      | 80 - 90	    | -10     |
| 90	      | 90 - 90	    |  0      |
| 100	      | 100 - 90	  | +10     |

# apply()  
* It is a equivalent method if we want to transform a whole DataFrame by calling a custom method on each row.

```python
# EX:
def mean_points (row):
  row.ponts = row.points - df_mean
  return row
df.apply(mean_points, axis = 'columns')
```

If we had called `reviews.apply()` with `axis='index'`, then instead of passing a function to transform each row, we would need to give a function to transform each* column*.
