# Coosing between `loc` and `iloc`
When choosing between `loc` and `iloc` there is one thing to remember.  
The 2 methods use slightly different indexing schemes.  
EX:


```
iloc[0:10] the output will be 0,1,2,...,9.

loc[0:10] the ouput will be 0,1,2,3....,10.
```



## **INDEX MANIPULATING**

````python
# EX:
df.set_index('ts')
````

## Conditional Selection
selecting the data by using the conditionals

Checking through the conditional operations.

````python
# EX: checking wether all source address or from 192.168.1.31
df.src_ip == '192.168.1.31'
````

This result can then be used inside of `loc` to select relevant data.

````python
df.loc[df.src_ip == '192.168.1.31']
````

````python
df.loc[(df.src_ip == '192.168.1.31') & (df.src_ip == 'udp')]
````



```
df.loc[(df.src_ip == '192.168.1.31') & (df.src_ip == 'udp')]
```
The above code tells that the if source ip should be `192.168.1.31` and and that should should be `udp`. if both conditions are `true` then it gives the output



OR Operator  

````python
# EX: OR, |
df.loc[(df.src_ip == '192.168.1.31') | (df.src_ip == 'udp')]
````

# Built-in conditional selectors in Pandas  
There are two important in those
 1. isin `isin()`
 2. isnull `isnull()`

# 1. isin  
* let's you select data whose value `is in` a list of values.

````python
df.loc[df.src_ip.isin(['192.168.1.31', '192.168.1.79'])]
````

# 2. isnull  
* lets you highlight values `which are empty (or not empty)`(NAN)

````python
# EX: 1
df.loc[df.proto.isnull()]
````

````python
# EX: 2
df.loc[df.proto.notnull()]
````
