Float: Floats are numbers with fractional parts. They can have many numbers after decimal.

```python
pyi_value = 3.141592653589793238462643383279502884197169399375105820974944
print(pyi_value)
print(type(pyi_value))
```

We can also specify a float with a fraction.

```python
pi = 22/7
print(pi)
print(type(pi))
```

Whenever we write a data with decimal python recognizes it as float

```python
y = 22.
print(y)
print(type(y))
```

Float has a function called "round()" to round a number to a specific number of decimals

```python
pie = round(pyi_value,3)
print(pie)
```

Boolean: TRUE/FALSE

```python
N = True
print(N)
type(N)
```

```python
# EX:2
Z = (2>1)
print(Z)
type(Z)
```

We can switch the value of a boolean by using not. So, not True is equivalent to False, and not False becomes True.

```python
Z1 = not True
print(Z1)
type(Z1)
```
